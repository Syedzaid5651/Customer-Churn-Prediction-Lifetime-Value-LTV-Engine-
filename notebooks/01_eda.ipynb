# 📊 Exploratory Data Analysis (EDA)

In this notebook, we will explore the **Telco Customer Churn** dataset.
Our goals are to:
1. Understand the data types and basic statistics
2. Handle any missing or messy data
3. Visualize the relationships between features and **Churn**
4. Identify the strongest indicators of churn for our ML models

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Set plotting style
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette('deep')

# Load the dataset
df = pd.read_csv("../data/raw/telco_customer_churn.csv")
print(f"Dataset Shape: {df.shape}")
df.head()
```

## 1. Data Cleaning
`TotalCharges` seems to be loaded as an object (string) because there are some blank spaces for new customers. We need to convert it to numeric and handle the missing values.

```python
# Convert TotalCharges to numeric (errors='coerce' will turn blank spaces into NaN)
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')

# Check how many NaNs we created
missing_charges = df['TotalCharges'].isnull().sum()
print(f"Missing TotalCharges values: {missing_charges}")

# Since they are new customers (tenure=0), their TotalCharges should be 0
df['TotalCharges'] = df['TotalCharges'].fillna(0)
print("Missing values filled with 0.")
```

## 2. Overall Churn Distribution
First, let's see how imbalanced our dataset is.

```python
plt.figure(figsize=(6, 4))
ax = sns.countplot(data=df, x='Churn', palette='viridis')
plt.title('Overall Churn Distribution')

# Add percentage labels
total = len(df)
for p in ax.patches:
    percentage = f'{100 * p.get_height() / total:.1f}%'
    x = p.get_x() + p.get_width() / 2
    y = p.get_height()
    ax.annotate(percentage, (x, y), ha='center', va='bottom', weight='bold')

plt.show()
```

## 3. Contract Type & Tenure vs Churn
Let's analyze if long-term customers are less likely to churn, and if contract type plays a big role.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Contract Type
sns.countplot(data=df, x='Contract', hue='Churn', ax=axes[0], palette='Set2')
axes[0].set_title('Churn by Contract Type')

# Tenure (Distribution)
sns.histplot(data=df, x='tenure', hue='Churn', ax=axes[1], multiple='stack', palette='Set2', bins=36)
axes[1].set_title('Churn by Tenure (Months)')

plt.tight_layout()
plt.show()
```

## 4. Monthly Charges vs Churn
Do people who pay more churn more? Let's visualize the distribution of Monthly Charges for churned vs retained customers.

```python
plt.figure(figsize=(8, 5))
sns.kdeplot(data=df, x='MonthlyCharges', hue='Churn', fill=True, common_norm=False, palette='Set1')
plt.title('Distribution of Monthly Charges (Churn vs Retained)')
plt.show()
```
